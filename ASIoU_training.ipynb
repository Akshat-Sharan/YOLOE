```python
from ultralytics import YOLOE
from ultralytics.models.yolo.yoloe import YOLOEPETrainer

# Initialize a detection model from a config
model = YOLOE("yoloe-26l.yaml")

# Load weights from a pretrained segmentation checkpoint (same scale)
model.load("yoloe-26l-seg.pt")

# Fine-tune on your detection dataset using your custom ASIoU loss
results = model.train(
    data="D:/AeroDefence/augmented/data.yaml",
    epochs=80,
    batch=50,
    patience=10,
    trainer=YOLOEPETrainer,
    project="ASIoU_Experiment",
    name="yoloe-26l-asiou_loss",
)
```
